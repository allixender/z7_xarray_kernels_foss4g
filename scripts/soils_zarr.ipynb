### Lightweight, practical usage examples for the Estonia soil IGEO7/Z7 DataTree.

These snippets demonstrate the day-to-day operations you need when working
with the converted archives:

- opening the DataTree and inspecting the DGGS index / grid metadata
- selecting cells by cell id
- selecting cells by latitude/longitude (a point)
- selecting cells by a bounding box
- deriving cell centers / boundaries
- interactive mapping (notebook only)

Everything here runs on ``res07`` (890 cells) by default so it stays fast; swap
``level=`` for a finer resolution when you want real work.

Run as a self-contained demo:

    pixi run python -m eesti_soil_conversion.samples

or import the individual functions:

    from eesti_soil_conversion.samples import select_by_bbox

```python
"""Return the ``clay_mean`` DataArray at ``level`` (lazy, dask-backed)."""
level = 8
dt = open_datatree(levels=(level,))
ds = dt[f"res{level:02d}"]
ds["clay_mean"]
```

```python
"""Show the DGGS index, grid metadata, and the cell-id coordinate."""
da= ds["clay_mean"]

idx = da.dggs.index
print(f"index    : {idx._repr_inline_(60)}")
print(f"grid     : {da.dggs.grid_info.to_dict()}")
cell_ids = da.dggs.cell_ids
print(f"cell_ids : {cell_ids.dims} {cell_ids.dtype} "
      f"first 3 = {cell_ids.values[:3]}")
```

```python
"""Select arbitrary cells by their packed Z7 ``uint64`` id.

Resolution is O(log R) against the range table — no dense coordinate is
materialised.
"""
cell_ids = np.asarray(cell_ids, dtype=np.uint64)
selected = da.sel(cell_ids=cell_ids)
selected
```

```python
"""Select the cell(s) containing the given latitude/longitude point(s).

``latitude``/``longitude`` are degrees (WGS84); each may be a scalar or an
array. Internally the point is projected to a Z7 cell id via the grid's
``geographic2cell_ids`` and then range-selected.
"""
latitude = [58.0208]
longitude = [26.6757]

lat = np.atleast_1d(np.asarray(latitude, dtype=float))
lon = np.atleast_1d(np.asarray(longitude, dtype=float))
selected = da.dggs.sel_latlon(latitude=lat, longitude=lon)
selected
```

```python
"""Select all cells whose centers fall inside a WGS84 bounding box.

A spatial mask is built from the cell centers, the matching cell ids are
collected, and a single range-aware ``sel`` fetches them.
"""
lat_min = latitude
lon_min = longitude

lat_max = 58.529
lon_max = 26.8971

centers = da.dggs.cell_centers()
lon = centers["longitude"].values
lat = centers["latitude"].values
mask = (lon >= lon_min) & (lon <= lon_max) & (lat >= lat_min) & (lat <= lat_max)
sub_ids = da.dggs.cell_ids.values[mask]
selected = da.sel(cell_ids=sub_ids)
selected
```

```python
"""Return cell centers as an ``xr.Dataset`` with ``latitude``/``longitude``."""
da.dggs.cell_centers()
```

```python
"""Return the hexagonal cell boundaries as shapely polygons."""
da.dggs.cell_boundaries()
```
